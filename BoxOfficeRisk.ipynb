````python
# ================================================================
# DAY 11: BOXOFFICERISK AI
# Movie Box Office Success Classification + Revenue Prediction
# Complete Google Colab Implementation
# ================================================================

# 1. INSTALL LIBRARIES
!pip -q install datasets scikit-learn pandas numpy matplotlib seaborn joblib

import os
import re
import json
import ast
import sqlite3
import warnings
import zipfile
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from datasets import load_dataset
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier
from sklearn.ensemble import RandomForestRegressor, ExtraTreesRegressor
from sklearn.dummy import DummyClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
from IPython.display import display

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")

RANDOM_STATE = 42

# 2. CREATE PROJECT FOLDER

PROJECT = "/content/BoxOfficeRisk_AI"

folders = [
    "outputs",
    "models",
    "reports",
    "data"
]

for folder in folders:
    os.makedirs(os.path.join(PROJECT, folder), exist_ok=True)

print("BoxOfficeRisk AI environment initialized.")

# 3. LOAD PUBLIC DATASET DIRECTLY

print("\nLoading TMDB Box Office dataset...")

dataset_id = "Jesscoyle111/TMDB-box-office"

try:
    dataset = load_dataset(dataset_id, split="train")
except Exception as first_error:
    print("Primary dataset unavailable. Trying alternate source...")
    dataset_id = "ashen97/TMDB-box-office"
    dataset = load_dataset(dataset_id, split="train")

df = dataset.to_pandas()

df.columns = [
    str(c).strip().lower().replace(" ", "_")
    for c in df.columns
]

print("Dataset loaded successfully!")
print("Source:", dataset_id)
print("Original shape:", df.shape)

display(df.head())

# 4. CHECK DATA QUALITY

print("\nDATA INFORMATION")
print(df.info())

print("\nMISSING VALUES")
display(
    df.isnull().sum()
       .sort_values(ascending=False)
       .to_frame("Missing Values")
)

print("\nDUPLICATE ROWS:", df.duplicated().sum())

# 5. VERIFY REQUIRED COLUMNS

required_columns = ["budget", "revenue", "release_date"]

missing_columns = [
    c for c in required_columns if c not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Dataset is missing required columns: {missing_columns}"
    )

# 6. CLEAN NUMERIC COLUMNS

for col in ["budget", "revenue", "runtime", "popularity",
            "vote_average", "vote_count"]:

    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

# Parse dates
df["release_date"] = pd.to_datetime(
    df["release_date"], errors="coerce"
)

# Remove duplicate movie records
if "id" in df.columns:
    df = df.drop_duplicates(subset=["id"])
else:
    df = df.drop_duplicates()

# Retain movies with a valid release date and positive budget/revenue
df = df.dropna(
    subset=["budget", "revenue", "release_date"]
).copy()

df = df[
    (df["budget"] > 0) &
    (df["revenue"] > 0) &
    (df["release_date"] <= pd.Timestamp.today())
].copy()

# Restrict to historically released records when status is available
if "status" in df.columns:
    released = df["status"].astype(str).str.lower().eq("released")
    if released.sum() > 0:
        df = df[released].copy()

if len(df) < 100:
    raise ValueError(
        "Too few valid movies after cleaning. Check the dataset source."
    )

df = df.reset_index(drop=True)

print("\nCLEANED DATASET SHAPE:", df.shape)

# 7. PARSE MOVIE METADATA

def extract_names(value):
    """
    Extract names from JSON-like metadata or plain text.
    Handles genre, cast and production company fields.
    """

    if value is None:
        return []

    if isinstance(value, float) and np.isnan(value):
        return []

    if isinstance(value, list):
        result = []
        for item in value:
            if isinstance(item, dict):
                name = item.get("name")
                if name:
                    result.append(str(name))
            elif item:
                result.append(str(item))
        return result

    text = str(value).strip()

    if not text or text.lower() in ["nan", "none", "null"]:
        return []

    try:
        parsed = json.loads(text)
        return extract_names(parsed)
    except Exception:
        pass

    try:
        parsed = ast.literal_eval(text)
        if isinstance(parsed, (list, tuple)):
            return extract_names(parsed)
    except Exception:
        pass

    # Fallback for comma-separated text
    return [
        item.strip(" []'\"")
        for item in text.split(",")
        if item.strip(" []'\"")
    ]

for col in ["genres", "cast", "production_companies"]:
    if col not in df.columns:
        df[col] = ""

df["genre_list"] = df["genres"].apply(extract_names)

df["primary_genre"] = df["genre_list"].apply(
    lambda x: x[0] if x else "Unknown"
)

df["cast_count"] = df["cast"].apply(
    lambda x: len(extract_names(x))
)

df["production_company_count"] = (
    df["production_companies"].apply(
        lambda x: len(extract_names(x))
    )
)

if "director" not in df.columns:
    df["director"] = "Unknown"

df["director"] = (
    df["director"].fillna("Unknown").astype(str)
)

df.loc[
    df["director"].str.lower().isin(["", "nan", "none"]),
    "director"
] = "Unknown"

if "original_language" not in df.columns:
    df["original_language"] = "Unknown"

df["original_language" ] = (
    df["original_language"].fillna("Unknown").astype(str)
)

# 8. FEATURE ENGINEERING

df["release_year"] = df["release_date"].dt.year
df["release_month"] = df["release_date"].dt.month

df["log_budget"] = np.log1p(df["budget"])

# Gross revenue-to-budget multiple.
# This is NOT net profit or true ROI.
df["gross_multiple"] = df["revenue"] / df["budget"]

# Avoid extreme numerical values from anomalous records.
df = df.replace([np.inf, -np.inf], np.nan)

# Define three outcome categories
def classify_movie(value):
    if value < 1:
        return "Flop"
    elif value < 2:
        return "Break-even / Moderate"
    return "Hit"

df["success_category"] = df["gross_multiple"].apply(
    classify_movie
)

print("\nSUCCESS CATEGORY DISTRIBUTION")
display(
    df["success_category"]
      .value_counts()
      .to_frame("Movies")
)

# 9. EXPLORATORY DATA ANALYSIS

print("\nNUMERICAL SUMMARY")
display(
    df[["budget", "revenue", "runtime",
        "popularity", "vote_average",
        "gross_multiple"]]
    .describe()
)

# 9A. Success category distribution

plt.figure(figsize=(9, 5))

order = ["Flop", "Break-even / Moderate", "Hit"]

sns.countplot(
    data=df,
    x="success_category",
    order=order
)

plt.title("Movie Box Office Outcome Distribution")
plt.xlabel("Outcome")
plt.ylabel("Number of Movies")
plt.xticks(rotation=15)
plt.tight_layout()

plt.savefig(
    f"{PROJECT}/outputs/success_distribution.png",
    dpi=150
)
plt.show()

# 9B. Budget distribution

plt.figure(figsize=(9, 5))

sns.histplot(
    np.log1p(df["budget"]),
    bins=40,
    kde=True
)

plt.title("Log-Transformed Movie Budget Distribution")
plt.xlabel("Log Budget")
plt.tight_layout()

plt.savefig(
    f"{PROJECT}/outputs/budget_distribution.png",
    dpi=150
)
plt.show()

# 9C. Revenue distribution

plt.figure(figsize=(9, 5))

sns.histplot(
    np.log1p(df["revenue"]),
    bins=40,
    kde=True
)

plt.title("Log-Transformed Box Office Revenue")
plt.xlabel("Log Revenue")
plt.tight_layout()

plt.savefig(
    f"{PROJECT}/outputs/revenue_distribution.png",
    dpi=150
)
plt.show()

# 9D. Genre performance

genre_analysis = (
    df.groupby("primary_genre")
      .agg(
          movie_count=("title", "count")
          if "title" in df.columns
          else ("budget", "count"),
          median_revenue=("revenue", "median"),
          median_budget=("budget", "median"),
          median_gross_multiple=("gross_multiple", "median"),
          average_gross_multiple=("gross_multiple", "mean")
      )
      .sort_values("movie_count", ascending=False)
)

print("\nGENRE ANALYSIS")
display(genre_analysis.head(15))

genre_analysis.to_csv(
    f"{PROJECT}/reports/genre_analysis.csv"
)

# 9E. Top genres by median gross multiple
# Apply a minimum sample size to reduce noisy comparisons.

reliable_genres = genre_analysis[
    genre_analysis["movie_count"] >= 10
].sort_values("median_gross_multiple", ascending=False)

plt.figure(figsize=(11, 6))

sns.barplot(
    data=reliable_genres.head(12).reset_index(),
    x="median_gross_multiple",
    y="primary_genre"
)

plt.title("Top Genres by Median Revenue-to-Budget Multiple")
plt.xlabel("Median Gross Multiple")
plt.ylabel("Genre")
plt.tight_layout()

plt.savefig(
    f"{PROJECT}/outputs/genre_performance.png",
    dpi=150
)
plt.show()

# 9F. Budget versus revenue

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="budget",
    y="revenue",
    hue="success_category",
    alpha=0.6
)

plt.xscale("log")
plt.yscale("log")
plt.title("Movie Budget vs Worldwide Revenue")
plt.tight_layout()

plt.savefig(
    f"{PROJECT}/outputs/budget_vs_revenue.png",
    dpi=150
)
plt.show()

# 9G. Correlation analysis

correlation_cols = [
    "budget", "revenue", "runtime",
    "popularity", "vote_average",
    "vote_count", "gross_multiple"
]

correlation_cols = [
    c for c in correlation_cols if c in df.columns
]

plt.figure(figsize=(10, 7))

sns.heatmap(
    df[correlation_cols].corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Movie Data Correlation Matrix")
plt.tight_layout()

plt.savefig(
    f"{PROJECT}/outputs/correlation_matrix.png",
    dpi=150
)
plt.show()

# 10. SQL ANALYSIS USING SQLITE

print("\nRUNNING SQL INDUSTRY ANALYSIS...")

# Drop lists and other un-serializable objects from the SQL slice
sql_df = df.copy()
sql_df = sql_df.drop(columns=["genre_list"], errors="ignore")

sql_df["release_year"] = sql_df["release_year"].astype(int)

db_path = f"{PROJECT}/data/boxoffice.db"

connection = sqlite3.connect(db_path)

sql_df.to_sql(
    "movies",
    connection,
    if_exists="replace",
    index=False
)

# Query 1: Top genres by movie count

query1 = """
SELECT
    primary_genre,
    COUNT(*) AS movie_count,
    ROUND(AVG(gross_multiple), 2) AS avg_gross_multiple
FROM movies
GROUP BY primary_genre
HAVING COUNT(*) >= 10
ORDER BY avg_gross_multiple DESC
"""

sql_genres = pd.read_sql_query(query1, connection)

print("\nSQL QUERY 1: GENRE PERFORMANCE")
display(sql_genres.head(15))

# Query 2: Annual movie revenue

query2 = """
SELECT
    release_year,
    COUNT(*) AS movie_count,
    ROUND(AVG(revenue), 0) AS average_revenue,
    ROUND(AVG(budget), 0) AS average_budget
FROM movies
GROUP BY release_year
ORDER BY release_year
"""

sql_yearly = pd.read_sql_query(query2, connection)

print("\nSQL QUERY 2: YEARLY INDUSTRY ANALYSIS")
display(sql_yearly.tail(20))

# Query 3: Success category analysis

query3 = """
SELECT
    success_category,
    COUNT(*) AS movie_count,
    ROUND(AVG(budget), 0) AS average_budget,
    ROUND(AVG(revenue), 0) AS average_revenue
FROM movies
GROUP BY success_category
ORDER BY movie_count DESC
"""

sql_success = pd.read_sql_query(query3, connection)

print("\nSQL QUERY 3: BOX OFFICE OUTCOMES")
display(sql_success)

sql_genres.to_csv(
    f"{PROJECT}/reports/sql_genre_analysis.csv",
    index=False
)

sql_yearly.to_csv(
    f"{PROJECT}/reports/sql_yearly_analysis.csv",
    index=False
)

sql_success.to_csv(
    f"{PROJECT}/reports/sql_success_analysis.csv",
    index=False
)

connection.close()

# 11. PREPARE MACHINE LEARNING FEATURES

# IMPORTANT:
# Do not use revenue, vote_average, vote_count or popularity
# as model inputs. They can contain post-release information.
# This project is intended to simulate pre-release prediction.

df["runtime"] = pd.to_numeric(
    df["runtime"], errors="coerce"
)

numeric_features = [
    "log_budget",
    "runtime",
    "release_month",
    "release_year",
    "cast_count",
    "production_company_count"
]

categorical_features = [
    "primary_genre",
    "original_language",
    "director"
]

# Limit director categories to avoid extremely sparse features.
top_directors = (
    df["director"].value_counts().head(30).index
)

df["director" ] = df["director"].where(
    df["director"].isin(top_directors),
    "Other"
)

feature_columns = numeric_features + categorical_features

X = df[feature_columns].copy()

y_class = df["success_category"].copy()
y_revenue = np.log1p(df["revenue"].copy())

# 12. TRAIN / TEST SPLIT

indices = np.arange(len(df))

class_counts = y_class.value_counts()

stratify_values = (
    y_class if class_counts.min() >= 2 else None
)

train_idx, test_idx = train_test_split(
    indices,
    test_size=0.20,
    random_state=RANDOM_STATE,
    stratify=stratify_values
)

X_train = X.iloc[train_idx]
X_test = X.iloc[test_idx]

y_train_class = y_class.iloc[train_idx]
y_test_class = y_class.iloc[test_idx]

y_train_revenue = y_revenue.iloc[train_idx]
y_test_revenue = y_revenue.iloc[test_idx]

print("\nTRAINING RECORDS:", len(X_train))
print("TESTING RECORDS:", len(X_test))

# 13. PREPROCESSING PIPELINE

numeric_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median"))
    ]
)

categorical_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="constant",
                          fill_value="Unknown")
        ),
        (
            "encoder",
            OneHotEncoder(handle_unknown="ignore")
        )
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)

# 14. CLASSIFICATION MODELS

classification_models = {
    "Random Forest": RandomForestClassifier(
        n_estimators=250,
        max_depth=18,
        min_samples_split=5,
        class_weight="balanced_subsample",
        random_state=RANDOM_STATE,
        n_jobs=-1
    ),

    "Extra Trees": ExtraTreesClassifier(
        n_estimators=250,
        max_depth=18,
        min_samples_split=5,
        class_weight="balanced",
        random_state=RANDOM_STATE,
        n_jobs=-1
    )
}

classification_results = {}
trained_classifiers = {}

for model_name, model in classification_models.items():

    print(f"\nTraining classifier: {model_name}")

    pipeline = Pipeline(
        steps=[
            ("preprocessor", preprocessor),
            ("model", model)
        ]
    )

    pipeline.fit(X_train, y_train_class)

    predictions = pipeline.predict(X_test)

    metrics = {
        "Model": model_name,
        "Accuracy": accuracy_score(
            y_test_class, predictions
        ),
        "Precision_Macro": precision_score(
            y_test_class, predictions,
            average="macro", zero_division=0
        ),
        "Recall_Macro": recall_score(
            y_test_class, predictions,
            average="macro", zero_division=0
        ),
        "F1_Macro": f1_score(
            y_test_class, predictions,
            average="macro", zero_division=0
        ),
        "F1_Weighted": f1_score(
            y_test_class, predictions,
            average="weighted", zero_division=0
        )
    }

    classification_results[model_name] = metrics
    trained_classifiers[model_name] = pipeline

    print(metrics)

# Baseline classifier

baseline = DummyClassifier(strategy="most_frequent")

baseline.fit(X_train, y_train_class)

baseline_predictions = baseline.predict(X_test)

baseline_f1 = f1_score(
    y_test_class,
    baseline_predictions,
    average="macro",
    zero_division=0
)

print("\nBASELINE MACRO F1:", baseline_f1)

classification_table = pd.DataFrame(
    classification_results.values()
)

classification_table = pd.concat(
    [
        classification_table,
        pd.DataFrame([{
            "Model": "Dummy Baseline",
            "Accuracy": accuracy_score(
                y_test_class, baseline_predictions
            ),
            "Precision_Macro": precision_score(
                y_test_class, baseline_predictions,
                average="macro", zero_division=0
            ),
            "Recall_Macro": recall_score(
                y_test_class, baseline_predictions,
                average="macro", zero_division=0
            ),
            "F1_Macro": baseline_f1,
            "F1_Weighted": f1_score(
                y_test_class, baseline_predictions,
                average="weighted", zero_division=0
            )
        }])
    ],
    ignore_index=True
)

classification_table = classification_table.sort_values(
    "F1_Macro", ascending=False
)

print("\nCLASSIFICATION MODEL COMPARISON")
display(classification_table)

classification_table.to_csv(
    f"{PROJECT}/reports/classification_model_comparison.csv",
    index=False
)

# Select best classifier using macro F1

best_classifier_name = max(
    trained_classifiers,
    key=lambda name: classification_results[name]["F1_Macro"]
)

best_classifier = trained_classifiers[best_classifier_name]

print("\nBEST CLASSIFIER:", best_classifier_name)

# 15. CLASSIFICATION REPORT

class_predictions = best_classifier.predict(X_test)

report = classification_report(
    y_test_class,
    class_predictions,
    output_dict=True,
    zero_division=0
)

report_df = pd.DataFrame(report).transpose()

display(report_df)

report_df.to_csv(
    f"{PROJECT}/reports/classification_report.csv"
)

# 16. CONFUSION MATRIX

class_labels = sorted(y_class.unique())

cm = confusion_matrix(
    y_test_class,
    class_predictions,
    labels=class_labels
)

plt.figure(figsize=(8, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=class_labels,
    yticklabels=class_labels
)

plt.title(f"Box Office Classification — {best_classifier_name}")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.xticks(rotation=20)
plt.yticks(rotation=0)
plt.tight_layout()

plt.savefig(
    f"{PROJECT}/outputs/confusion_matrix.png",
    dpi=150
)
plt.show()

# 17. REVENUE REGRESSION MODELS

regression_models = {
    "Random Forest Regressor": RandomForestRegressor(
        n_estimators=250,
        max_depth=18,
        min_samples_split=5,
        random_state=RANDOM_STATE,
        n_jobs=-1
    ),

    "Extra Trees Regressor": ExtraTreesRegressor(
        n_estimators=250,
        max_depth=18,
        min_samples_split=5,
        random_state=RANDOM_STATE,
        n_jobs=-1
    )
}

regression_results = {}
trained_regressors = {}

for model_name, model in regression_models.items():

    print(f"\nTraining regressor: {model_name}")

    pipeline = Pipeline(
        steps=[
            ("preprocessor", preprocessor),
            ("model", model)
        ]
    )

    pipeline.fit(X_train, y_train_revenue)

    log_predictions = pipeline.predict(X_test)

    actual_revenue = np.expm1(
        y_test_revenue.to_numpy()
    )

    predicted_revenue = np.maximum(
        np.expm1(log_predictions), 0
    )

    mae = mean_absolute_error(
        actual_revenue, predicted_revenue
    )

    rmse = np.sqrt(
        mean_squared_error(
            actual_revenue, predicted_revenue
        )
    )

    r2 = r2_score(
        actual_revenue, predicted_revenue
    )

    # RMSLE, computed on nonnegative values
    rmsle = np.sqrt(
        mean_squared_error(
            np.log1p(actual_revenue),
            np.log1p(predicted_revenue)
        )
    )

    metrics = {
        "Model": model_name,
        "MAE": mae,
        "RMSE": rmse,
        "R2": r2,
        "RMSLE": rmsle
    }

    regression_results[model_name] = metrics
    trained_regressors[model_name] = pipeline

    print(metrics)

regression_table = pd.DataFrame(
    regression_results.values()
).sort_values("RMSLE")

print("\nREGRESSION MODEL COMPARISON")
display(regression_table)

regression_table.to_csv(
    f"{PROJECT}/reports/regression_model_comparison.csv",
    index=False
)

best_regressor_name = regression_table.iloc[0]["Model"]

best_regressor = trained_regressors[best_regressor_name]

print("\nBEST REGRESSOR:", best_regressor_name)

# 18. REGRESSION VISUALIZATION

log_predictions = best_regressor.predict(X_test)

actual_revenue = np.expm1(
    y_test_revenue.to_numpy()
)

predicted_revenue = np.maximum(
    np.expm1(log_predictions), 0
)

plt.figure(figsize=(9, 6))

plt.scatter(
    actual_revenue,
    predicted_revenue,
    alpha=0.5
)

upper = max(
    float(np.max(actual_revenue)),
    float(np.max(predicted_revenue))
)

plt.plot(
    [0, upper],
    [0, upper],
    linestyle="--"
)

plt.xscale("symlog")
plt.yscale("symlog")

plt.title("Actual vs Predicted Movie Revenue")
plt.xlabel("Actual Revenue")
plt.ylabel("Predicted Revenue")
plt.tight_layout()

plt.savefig(
    f"{PROJECT}/outputs/revenue_prediction.png",
    dpi=150
)
plt.show()

# 19. FEATURE IMPORTANCE

def plot_feature_importance(pipeline, filename, title):

    preprocessor_fitted = pipeline.named_steps["preprocessor"]
    model_fitted = pipeline.named_steps["model"]

    feature_names = preprocessor_fitted.get_feature_names_out()

    importance = model_fitted.feature_importances_

    importance_df = pd.DataFrame({
        "Feature": feature_names,
        "Importance": importance
    }).sort_values(
        "Importance", ascending=False
    ).head(15)

    display(importance_df)

    plt.figure(figsize=(10, 7))

    sns.barplot(
        data=importance_df,
        x="Importance",
        y="Feature"
    )

    plt.title(title)
    plt.tight_layout()

    plt.savefig(
        filename,
        dpi=150
    )

    plt.show()

    return importance_df

classifier_importance = plot_feature_importance(
    best_classifier,
    f"{PROJECT}/outputs/classifier_feature_importance.png",
    "Classification Feature Importance"
)

regressor_importance = plot_feature_importance(
    best_regressor,
    f"{PROJECT}/outputs/regression_feature_importance.png",
    "Revenue Regression Feature Importance"
)

classifier_importance.to_csv(
    f"{PROJECT}/reports/classifier_feature_importance.csv",
    index=False
)

regressor_importance.to_csv(
    f"{PROJECT}/reports/regression_feature_importance.csv",
    index=False
)

# 20. SAVE TRAINED MODELS

joblib.dump(
    best_classifier,
    f"{PROJECT}/models/boxoffice_classifier.pkl"
)

joblib.dump(
    best_regressor,
    f"{PROJECT}/models/revenue_regressor.pkl"
)

joblib.dump(
    {
        "numeric_features": numeric_features,
        "categorical_features": categorical_features,
        "feature_columns": feature_columns,
        "top_directors": list(top_directors),
        "success_thresholds": {
            "Flop": "gross multiple < 1",
            "Break-even / Moderate": "1 <= gross multiple < 2",
            "Hit": "gross multiple >= 2"
        },
        "classifier": best_classifier_name,
        "regressor": best_regressor_name
    },
    f"{PROJECT}/models/model_metadata.pkl"
)

print("\nMODELS SAVED SUCCESSFULLY!")

# 21. INTERACTIVE MOVIE PREDICTION

def predict_movie(
    budget,
    runtime,
    release_month,
    release_year,
    genre,
    language="en",
    director="Unknown",
    cast_count=5,
    production_company_count=1
):

    if budget <= 0:
        raise ValueError("Budget must be greater than zero.")

    if runtime <= 0:
        raise ValueError("Runtime must be greater than zero.")

    if not 1 <= release_month <= 12:
        raise ValueError("Release month must be between 1 and 12.")

    if release_year < 1900:
        raise ValueError("Please enter a valid release year.")

    director = str(director)

    if director not in top_directors:
        director = "Other"

    movie_input = pd.DataFrame([{
        "log_budget": np.log1p(budget),
        "runtime": runtime,
        "release_month": release_month,
        "release_year": release_year,
        "cast_count": cast_count,
        "production_company_count": production_company_count,
        "primary_genre": genre,
        "original_language": language,
        "director": director
    }])

    predicted_class = best_classifier.predict(
        movie_input
    )[0]

    predicted_revenue_log = best_regressor.predict(
        movie_input
    )[0]

    predicted_revenue = max(
        0, float(np.expm1(predicted_revenue_log))
    )

    predicted_multiple = predicted_revenue / budget

    if hasattr(best_classifier, "predict_proba"):
        probabilities = best_classifier.predict_proba(
            movie_input
        )[0]

        probability_df = pd.DataFrame({
            "Category": best_classifier.classes_,
            "Model Probability": probabilities
        }).sort_values(
            "Model Probability", ascending=False
        )
    else:
        probability_df = pd.DataFrame()

    print("\n" + "=" * 55)
    print("             BOXOFFICERISK AI REPORT")
    print("=" * 55)

    print(f"Estimated Revenue: {predicted_revenue:,.0f}")
    print(f"Estimated Gross Multiple: {predicted_multiple:.2f}x")
    print(f"Predicted Category: {predicted_class}")

    print("\nCLASSIFICATION MODEL PROBABILITIES")
    display(probability_df)

    print("\nIMPORTANT:")
    print("These are model estimates, not guaranteed outcomes.")
    print("This model has not been validated for investment decisions.")
    print("=" * 55)

    return {
        "predicted_category": predicted_class,
        "estimated_revenue": predicted_revenue,
        "estimated_gross_multiple": predicted_multiple,
        "probabilities": probability_df
    }

example_prediction = predict_movie(
    budget=50000000,
    runtime=140,
    release_month=12,
    release_year=2026,
    genre="Action",
    language="en",
    director="Unknown",
    cast_count=6,
    production_company_count=2
)

# 22. EVALUATION SUMMARY

summary = {
    "dataset": dataset_id,
    "cleaned_records": int(len(df)),
    "training_records": int(len(train_idx)),
    "testing_records": int(len(test_idx)),
    "best_classifier": best_classifier_name,
    "best_regressor": best_regressor_name,
    "classification_metrics": classification_results[
        best_classifier_name
    ],
    "regression_metrics": regression_results[
        best_regressor_name
    ],
    "baseline_macro_f1": float(baseline_f1)
}

with open(
    f"{PROJECT}/reports/evaluation_summary.json",
    "w"
) as file:
    json.dump(summary, file, indent=4)

# 23. CREATE REQUIREMENTS FILE

requirements = """datasets
pandas
numpy
matplotlib
seaborn
scikit-learn
joblib
"""

with open(
    f"{PROJECT}/requirements.txt",
    "w"
) as file:
    file.write(requirements)

# 24. CREATE README.MD

readme = r"""
# BoxOfficeRisk AI
### Movie Box Office Success Classification and Revenue Prediction

## Overview

BoxOfficeRisk AI is a machine learning project designed to investigate
historical movie box office performance.

It combines two predictive systems:

1. Movie success classification.
2. Box office revenue regression.

It also includes exploratory analysis, SQL analytics, feature engineering,
model comparison, feature importance and interactive predictions.

## Problem Statement

Film production involves financial uncertainty. A movie's budget, genre,
release timing, runtime and production characteristics may be associated
with its historical box office performance.

This project investigates those relationships and develops machine
learning models using historical movie metadata.

## Dataset

TMDB Box Office Dataset:
https://huggingface.co/datasets/Jesscoyle111/TMDB-box-office

The dataset contains movie metadata including budget, revenue, genres,
runtime, release date, production companies, cast and director.

Dataset license: CC-BY-NC-SA 4.0 as indicated by the dataset card.
Review its terms before reuse or redistribution.

The project uses the Hugging Face datasets library to load the data
directly into Google Colab.

## Project Objectives

- Analyse historical box office performance.
- Identify genre-level revenue patterns.
- Explore budget and revenue relationships.
- Perform SQL-based movie industry analysis.
- Build a movie success classification model.
- Build a revenue prediction model.
- Compare multiple machine learning algorithms.
- Interpret model feature importance.
- Provide an interactive prediction function.

## Machine Learning Tasks

### 1. Classification

Three outcome classes are defined using the gross
revenue-to-budget multiple:

- Flop: below 1x
- Break-even / Moderate: 1x to below 2x
- Hit: 2x or higher

These are project-defined analytical categories, not official
industry classifications.

### 2. Regression

Predicts log-transformed worldwide box office revenue and converts
the prediction back into the dataset's revenue units.

## Feature Engineering

Features include:

- Log-transformed budget
- Runtime
- Release month
- Release year
- Cast count
- Production company count
- Primary genre
- Original language
- Director category

Revenue, popularity, ratings and vote counts are not used as
pre-release model inputs.

## Algorithms

Classification:
- Dummy Classifier
- Random Forest Classifier
- Extra Trees Classifier

Regression:
- Random Forest Regressor
- Extra Trees Regressor

## Evaluation Metrics

Classification:
- Accuracy
- Macro Precision
- Macro Recall
- Macro F1
- Weighted F1
- Confusion Matrix

Regression:
- Mean Absolute Error
- Root Mean Squared Error
- R-squared
- Root Mean Squared Logarithmic Error

Actual results are recorded in reports/evaluation_summary.json
and the model comparison CSV files.

## SQL Analysis

SQLite is used to investigate:

- Genre performance
- Yearly movie revenue patterns
- Budget and revenue by outcome category

## Visualizations

The outputs folder contains:

- Movie outcome distribution
- Budget distribution
- Revenue distribution
- Genre performance
- Budget versus revenue
- Correlation matrix
- Classification confusion matrix
- Revenue prediction comparison
- Classification feature importance
- Regression feature importance

## Repository Structure

```text
BoxOfficeRisk-AI/
|
|-- BoxOfficeRisk_AI_Day11.ipynb
|-- outputs/
|-- reports/
|-- models/
|-- data/
|-- requirements.txt
|-- README.md
|-- .gitignore
```

## How to Run

1. Open the notebook in Google Colab.
2. Run the complete implementation.
3. The dataset loads automatically through Hugging Face.
4. Review the analysis and model evaluation.
5. Test the interactive prediction function.

## Important Limitations

- Historical associations do not establish causation.
- Gross revenue divided by production budget is not net profit.
- Marketing, distribution, cinema shares, inflation and regional
  differences are not fully accounted for.
- Historical data may not represent today's film market.
- Random train/test splitting does not establish future-year
  forecasting performance.
- A director's identity and production characteristics may be
  associated with historical patterns that do not generalize.
- Model estimates should not be used as investment advice.
- Classification probabilities are model outputs, not guaranteed
  real-world probabilities.

## Future Improvements

- Temporal train/test validation.
- Inflation-adjusted revenue modelling.
- Indian cinema-specific analysis.
- Production budget and marketing cost modelling.
- SHAP-based explainability.
- Streamlit deployment.
- More robust data validation and monitoring.

## Author

Aditya Dhage

Data Science and AI Portfolio Project

## Disclaimer

This is an educational machine learning prototype, not a
production-grade financial forecasting system.

Movie data is associated with TMDB. This project is not endorsed
or certified by TMDB.
"""

with open(
    f"{PROJECT}/README.md",
    "w",
    encoding="utf-8"
) as file:
    file.write(readme.strip())

# 25. CREATE .GITIGNORE

gitignore = """
__pycache__/
.ipynb_checkpoints/
.env
*.pyc
.DS_Store
"""

with open(
    f"{PROJECT}/.gitignore",
    "w"
) as file:
    file.write(gitignore.strip())

# 26. SAVE CLEANED DATA

df.drop(
    columns=["genre_list"],
    errors="ignore"
).to_csv(
    f"{PROJECT}/data/cleaned_movies.csv",
    index=False
)

# 27. CREATE ZIP FILE

zip_path = "/content/BoxOfficeRisk_AI_Day11.zip"

with zipfile.ZipFile(
    zip_path, "w", zipfile.ZIP_DEFLATED
) as zipf:

    for root, dirs, files in os.walk(PROJECT):
        for filename in files:
            filepath = os.path.join(root, filename)
            arcname = os.path.relpath(filepath, PROJECT)
            zipf.write(filepath, arcname)

print("\n" + "=" * 65)
print("              BOXOFFICERISK  COMPLETE")
print("=" * 65)

print("Cleaned movies:", len(df))
print("Best classifier:", best_classifier_name)
print("Best regressor:", best_regressor_name)

print("\nCLASSIFICATION RESULTS")
display(classification_table)

print("\nREGRESSION RESULTS")
display(regression_table)

print("\nPROJECT DIRECTORY:", PROJECT)
print("ZIP FILE:", zip_path)

print("\nIMPORTANT: its done nowwwww.")

# Download the complete project package
from google.colab import files
files.download(zip_path)
````
